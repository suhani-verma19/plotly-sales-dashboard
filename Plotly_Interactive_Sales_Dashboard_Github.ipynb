```python
#Step 1: Import Libraries

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

#Step 2: Create Dataset

sales_df=pd.DataFrame({
    "Month":[
        "Jan","Feb","Mar","Apr",
        "May","Jun","Jul","Aug",
        "Sep","Oct","Nov","Dec"
    ],
    "Sales": [
        25000, 28000, 22000, 35000,
        40000, 38000, 45000, 48000,
        42000, 55000, 60000, 68000
    ],

    "Profit": [
        5000, 6000, 4000, 8000,
        10000, 9000, 12000, 13000,
        11000, 15000, 17000, 20000
    ],

    "Orders": [
        120, 135, 110, 160,
        180, 175, 200, 215,
        190, 240, 260, 290
    ]

})

#Step 3:Display Dataset
print("Sales Dashboard Dataset")
display(sales_df)

#Step 4:Dataset Information

print("Dataset Shape:", sales_df.shape)
print("\nDataset Information:")
sales_df.info()
```

# Interactive Sales & Profit Dashboard

## Project Overview
This project analyzes monthly sales, profit, and orders using Python, Pandas, and Plotly.

**Note:** The data in this notebook is sample data created for learning purposes. It is not real business data.

## Objective
- Analyze monthly sales and profit trends.
- Compare business performance.
- Build interactive visualizations.
- Identify important business insights.

## Tools Used
- Python
- Pandas
- Plotly
- Google Colab

```python
#Exploratory Data Analysis
#1. Check Missing values

print("Missing Values:")
print(sales_df.isnull().sum())

#2. Check duplicate rows
print("\nDuplicate Rows:")
print(sales_df.duplicated().sum())

#3. Statistical summary
print("\nStatistical Summary:")
display(sales_df.describe())

#4. Total sales&profit
print("\nTotal Sales:", sales_df["Sales"].sum())
print("Total Profit:",sales_df["Profit"].sum())

#5. Average Sales&Profit
print("\nAverage Sales:",sales_df["Sales"].mean())
print("Average Profit:",sales_df["Profit"].mean())

#6. Highest&Lowest sales
print("\nHighest Sales:",sales_df["Sales"].max())
print("Lowest Sales:",sales_df["Sales"].min())
```

```python
#Monthly Sales Trend Analysis
fig=px.line(
    sales_df,
    x="Month",
    y="Sales",
    markers=True,
    title="Monthly Sales Trend"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales (₹)",
    title_x=0.5
)

fig.update_traces(
    line=dict(width=3),
    marker=dict(size=10)
)

fig.show()
```

```python
#Monthly Profit Analysis
fig=px.bar(
    sales_df,
    x="Month",
    y="Profit",
    color="Profit",
    text="Profit",
    title="Monthly Profit Analysis"
)
fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Profit (₹)",
    title_x=0.5
)
fig.update_traces(
    texttemplate="₹%{y:,}",
    textposition="outside"
)

fig.show()
```

```python
# Sales vs Profit Comparison
fig=go.Figure()
fig.add_trace(
    go.Bar(
        x=sales_df["Month"],
        y=sales_df["Sales"],
        name="Sales",
        marker_color="royalblue"
    )
)
fig.add_trace(
    go.Bar(
        x=sales_df["Month"],
        y=sales_df["Profit"],
        name="Profit",
        marker_color="orange"
    )
)
fig.update_layout(
    title="Monthly Sales Vs Profit Comparison",
    xaxis_title="Month",
    yaxis_title="Amount (₹)",
    barmode="group",
    title_x=0.5,
    hovermode="x unified"
)
fig.show()
```

```python

# Interactive Sales and Profit Dropdown

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=sales_df["Month"],
        y=sales_df["Sales"],
        name="Sales",
        marker_color="royalblue",
        visible=True
    )
)

fig.add_trace(
    go.Bar(
        x=sales_df["Month"],
        y=sales_df["Profit"],
        name="Profit",
        marker_color="orange",
        visible=False
    )
)

fig.update_layout(
    title="Monthly Sales Analysis",
    xaxis_title="Month",
    yaxis_title="Sales (₹)",
    title_x=0.5,
    updatemenus=[
        dict(
            buttons=[
                dict(
                    label="Sales",
                    method="update",
                    args=[
                        {"visible": [True, False]},
                        {
                            "title": "Monthly Sales Analysis",
                            "yaxis.title": "Sales (₹)"
                        }
                    ]
                ),
                dict(
                    label="Profit",
                    method="update",
                    args=[
                        {"visible": [False, True]},
                        {
                            "title": "Monthly Profit Analysis",
                            "yaxis.title": "Profit (₹)"
                        }
                    ]
                )
            ],
            direction="down",
            x=0.1,
            y=1.15
        )
    ]
)

fig.show()
```

```python
#Monthly Orders Analysis

fig=px.bar(
    sales_df,
    x="Month",
    y="Orders",
    color="Orders",
    text="Orders",
    title="Monthly Orders Analysis"
)
fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Number of Orders",
    title_x=0.5
)
fig.update_traces(
    textposition="outside"
)
fig.show()
```

```python

# Calculate Monthly Profit Margin

sales_df["Profit_Margin"] = (
    sales_df["Profit"] / sales_df["Sales"]
) * 100

display(
    sales_df[["Month", "Sales", "Profit", "Profit_Margin"]]
)
```

```python
#Monthly Profit Margin Chart
fig=px.line(
    sales_df,
    x="Month",
    y="Profit_Margin",
    markers=True,
    title="Monthly Profit Margin Trend"

)
fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Profit Margin (%)",
    title_x=0.5
)
fig.update_traces(
    line=dict(width=3),
    marker=dict(size=9),
    hovertemplate="Month: %{x}<br>Profit Margin: %{y:.2f}%<extra></extra>"
)
fig.show()
```

```python
#Calculate Average Order Value

sales_df["Average_Order_Value"]=(
    sales_df["Sales"] / sales_df["Orders"]
)

display(
    sales_df[[
        "Month","Sales","Orders","Average_Order_Value"
    ]]
)
```

```python
#Monthly Average Order Value Chart

fig=px.line(
    sales_df,
    x="Month",
    y="Average_Order_Value",
    markers=True,
    title="Monthly average order value"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Average Order Value (₹)",
    title_x=0.5
)

fig.update_traces(
    line=dict(width=3),
    marker=dict(size=9),
    hovertemplate="Month: %{x}<br>AOV: ₹%{y:.2f}<extra></extra>"

)

fig.show()
```

```python

from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Calculate business KPIs

total_sales = sales_df["Sales"].sum()
total_profit = sales_df["Profit"].sum()
total_orders = sales_df["Orders"].sum()

overall_profit_margin = (
    total_profit / total_sales
) * 100

overall_aov = total_sales / total_orders


# Create KPI Dashboard

fig = make_subplots(
    rows=1,
    cols=5,
    specs=[[{"type": "indicator"}] * 5]
)

# Total Sales
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_sales,
        title={"text": "Total Sales"},
        number={"prefix": "₹", "valueformat": ",.0f"}
    ),
    row=1, col=1
)

# Total Profit
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_profit,
        title={"text": "Total Profit"},
        number={"prefix": "₹", "valueformat": ",.0f"}
    ),
    row=1, col=2
)

# Total Orders
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_orders,
        title={"text": "Total Orders"},
        number={"valueformat": ",.0f"}
    ),
    row=1, col=3
)

# Overall Profit Margin
fig.add_trace(
    go.Indicator(
        mode="number",
        value=overall_profit_margin,
        title={"text": "Profit Margin"},
        number={"suffix": "%", "valueformat": ".2f"}
    ),
    row=1, col=4
)

# Average Order Value
fig.add_trace(
    go.Indicator(
        mode="number",
        value=overall_aov,
        title={"text": "Average Order Value"},
        number={"prefix": "₹", "valueformat": ".2f"}
    ),
    row=1, col=5
)

fig.update_layout(
    title={
        "text": "Business Performance Overview",
        "x": 0.5
    },
    height=300
)

fig.show()
```

```python

from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Create a 2x2 dashboard layout

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        "Monthly Sales Trend",
        "Monthly Profit Analysis",
        "Monthly Orders Analysis",
        "Profit Margin Trend"
    )
)

# Chart 1: Sales Trend
fig.add_trace(
    go.Scatter(
        x=sales_df["Month"],
        y=sales_df["Sales"],
        mode="lines+markers",
        name="Sales"
    ),
    row=1, col=1
)

# Chart 2: Profit Analysis
fig.add_trace(
    go.Bar(
        x=sales_df["Month"],
        y=sales_df["Profit"],
        name="Profit"
    ),
    row=1, col=2
)

# Chart 3: Orders Analysis
fig.add_trace(
    go.Bar(
        x=sales_df["Month"],
        y=sales_df["Orders"],
        name="Orders"
    ),
    row=2, col=1
)

# Chart 4: Profit Margin Trend
fig.add_trace(
    go.Scatter(
        x=sales_df["Month"],
        y=sales_df["Profit_Margin"],
        mode="lines+markers",
        name="Profit Margin"
    ),
    row=2, col=2
)

# Update dashboard layout
fig.update_layout(
    title={
        "text": "Interactive Business Performance Dashboard",
        "x": 0.5
    },
    height=750,
    showlegend=True
)

fig.show()
```

```python

# Calculate key business metrics

total_sales = sales_df["Sales"].sum()
total_profit = sales_df["Profit"].sum()
total_orders = sales_df["Orders"].sum()

overall_profit_margin = (
    total_profit / total_sales
) * 100

overall_aov = total_sales / total_orders

# Identify best and lowest performing months

highest_sales_month = sales_df.loc[
    sales_df["Sales"].idxmax(), "Month"
]

lowest_sales_month = sales_df.loc[
    sales_df["Sales"].idxmin(), "Month"
]

highest_profit_month = sales_df.loc[
    sales_df["Profit"].idxmax(), "Month"
]

highest_orders_month = sales_df.loc[
    sales_df["Orders"].idxmax(), "Month"
]

highest_aov_month = sales_df.loc[
    sales_df["Average_Order_Value"].idxmax(), "Month"
]

# Display business summary

print("BUSINESS PERFORMANCE SUMMARY")
print("-" * 35)

print(f"Total Sales: ₹{total_sales:,.0f}")
print(f"Total Profit: ₹{total_profit:,.0f}")
print(f"Total Orders: {total_orders:,}")
print(f"Overall Profit Margin: {overall_profit_margin:.2f}%")
print(f"Overall AOV: ₹{overall_aov:.2f}")

print("\nBUSINESS INSIGHTS")
print("-" * 35)

print(f"Highest Sales Month: {highest_sales_month}")
print(f"Lowest Sales Month: {lowest_sales_month}")
print(f"Highest Profit Month: {highest_profit_month}")
print(f"Highest Orders Month: {highest_orders_month}")
print(f"Highest AOV Month: {highest_aov_month}")
```
